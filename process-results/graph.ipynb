```python
def fetchRow(group):
    return df_combined[
        (df_combined['vpn'] == group[0]) &
        (df_combined['delay'] == group[1]) &
        (df_combined['jitter'] == group[2]) &
        (df_combined['loss'] == group[3]) &
        (df_combined['cpu_freq'] == group[4]) &
        (df_combined['core_count'] == group[5])
    ].iloc[0]

gdf = df_flent.groupby(key_cols)
for group, frame in gdf:
    plt.figure(figsize=(10, 6))
    plt.title(f"Group: {group}")
    for s in frame["down"]:
        row = fetchRow(group)
        ax = sns.lineplot(data=s)
        ax.axhline(y=row["down_mean"], color='red', linestyle='-', linewidth=0.5)
    plt.show()
```
