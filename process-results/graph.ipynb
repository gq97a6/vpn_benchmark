```python

import seaborn as sns
import matplotlib.pyplot as plt
from utils import parse_benchmark_to_dataframe
from static import *
```

```python
df, df_mean, df_sd, df_rsd, df_combined, df_flent = parse_benchmark_to_dataframe(project_dir / "results")
```

```python
gdf = df_flent.groupby(key_cols)

for group, frame in gdf:
    n = len(frame)
    cols = 5 
    rows = 2
    
    # squeeze=False ensures axes is always a 2D array, safely flattenable
    fig, axes = plt.subplots(rows, cols, figsize=(20, 5), squeeze=False)
    axes = axes.flatten()
    
    for i, s in enumerate(frame["tcp_down"]):
        sns.lineplot(data=s, ax=axes[i])
        
    # Remove empty subplots if n isn't a perfect multiple of cols
    for j in range(n, len(axes)):
        fig.delaxes(axes[j])
        
    fig.suptitle(f"Group: {group}")
    plt.tight_layout()
    plt.show()

    # Distribution (Histogram/KDE)
    #sns.histplot(s, kde=True)
    #plt.show()

    # Categorical counts / Bar plot
    #sns.barplot(x=s.index, y=s.values)
```
